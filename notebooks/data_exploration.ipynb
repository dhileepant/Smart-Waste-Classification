# 📊 Exploratory Data Analysis (EDA) - Smart Waste Classification
This notebook explores the dataset distribution, image resolutions, color channel distributions, and sample visualizer for the **Hazardous** and **Recyclable** waste categories.

```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

%matplotlib inline
sns.set_theme(style="darkgrid")
```

## 1. Dataset Directory & Class Counts

```python
data_dir = "../data"
classes = [d for d in os.listdir(data_dir) if os.path.isdir(os.path.join(data_dir, d))]

class_counts = {}
for cls in classes:
    class_path = os.path.join(data_dir, cls)
    count = len([f for f in os.listdir(class_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
    class_counts[cls] = count

df_counts = pd.DataFrame(list(class_counts.items()), columns=["Category", "Sample Count"])
print(df_counts)

# Visualize class distribution
plt.figure(figsize=(8, 5))
palette = ["#ef4444", "#10b981"]
bars = plt.bar(df_counts["Category"], df_counts["Sample Count"], color=palette, width=0.5)
plt.title("Dataset Sample Distribution", fontsize=14, fontweight="bold")
plt.ylabel("Number of Images", fontsize=12)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 20, int(yval), ha='center', va='bottom', fontweight='bold')
plt.show()
```

## 2. Sample Image Gallery for Each Waste Category

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle("Sample Waste Images (OpenCV Preprocessed)", fontsize=16, fontweight="bold")

for row_idx, cls in enumerate(classes):
    class_path = os.path.join(data_dir, cls)
    sample_files = os.listdir(class_path)[:4]
    for col_idx, file_name in enumerate(sample_files):
        img_path = os.path.join(class_path, file_name)
        img_bgr = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        img_resized = cv2.resize(img_rgb, (224, 224))
        
        ax = axes[row_idx, col_idx]
        ax.imshow(img_resized)
        ax.set_title(f"{cls.capitalize()}\n({img_bgr.shape[1]}x{img_bgr.shape[0]})")
        ax.axis("off")

plt.tight_layout()
plt.show()
```

## 3. Color Channel Pixel Intensity Distribution

```python
# Sample pixel distribution from recyclable and hazardous items
sample_rec_path = os.path.join(data_dir, "recyclable", os.listdir(os.path.join(data_dir, "recyclable"))[0])
img_sample = cv2.imread(sample_rec_path)
img_sample_rgb = cv2.cvtColor(img_sample, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 4))
colors = ('red', 'green', 'blue')
for i, color in enumerate(colors):
    hist = cv2.calcHist([img_sample_rgb], [i], None, [256], [0, 256])
    plt.plot(hist, color=color, label=f'{color.capitalize()} channel', linewidth=1.5)

plt.title("RGB Color Histogram of Representative Waste Item", fontsize=13, fontweight="bold")
plt.xlabel("Pixel Intensity [0 - 255]")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
